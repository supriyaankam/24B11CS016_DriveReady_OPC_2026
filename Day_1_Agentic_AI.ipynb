```python
from langchain_groq import ChatGroq
import os
from google.colab import userdata

os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)
```

```python
SYSTEM = """
You are CampusPulse, an AI Student Opportunity Assistant.

Your purpose is to help college students understand
internships, workshops, scholarships and hackathons.

Your responsibilities:
1. Identify the opportunity type.
2. Extract eligibility criteria.
3. Find the registration deadline.
4. Identify required documents.
5. Compare the opportunity with the student profile.
6. Create a simple action plan.

Rules:
- Use only the information provided.
- Never invent deadlines or eligibility requirements.
- If information is missing, say "Not mentioned".
- Clearly explain whether the student appears eligible.
- Use simple English.
- Give the response in an organized format.
"""
```

```python
def ask(question):
    return llm.invoke([
        ("system", SYSTEM),
        ("human", question)
    ]).content
```

```python
student_profile = """
I am a 2nd-year B.Tech CSE student.
I know basic Python.
I am interested in internships.
"""

announcement = """
ABC Technologies is offering a Python internship
for 2nd and 3rd-year CSE students.

Applicants must have basic Python knowledge.

Students must submit their resume before December 5.

Registration is through the college placement portal.
"""

question = f"""
Student Profile:
{student_profile}

Announcement:
{announcement}

Analyze this opportunity and provide:
- Opportunity name
- Eligibility
- Deadline
- Required documents
- Eligibility match
- Action plan
"""

print(ask(question))
```
